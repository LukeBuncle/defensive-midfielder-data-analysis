```python
import pandas as pd

# Convert the data into a list of dictionaries to make it easier to interpret 

data = []

for rd in raw_data:
    data.append({'Name':rd[0],
            'Age': rd[1],
            'Height': rd[2],
            'Minutes played (League)': rd[3],
            'Key passes per 90': rd[4],
            'Passing accuracy (%)': rd[5].replace('%',''),
            'Long pass accuracy (%)': rd[6].replace('%',''),
            'Interceptions per 90': rd[7],
            'Tackles per 90': rd[8],
            'Balls recovered per game': rd[9],
            'Total duels won per game': rd[10][0],
            'Duels won (%)': rd[10][1].replace('%',''),
            'Errors leading to goal or shot': rd[11],
            'Errors leading to goal or shot per 90': rd[12],
            'Successful dribbles per 90': rd[13][0],
            'Successful dribbles per 90 (%)': rd[13][1].replace('%',''),
            'Aerial duels won per 90': rd[14][0],
            'Aerial duels won per 90 (%)': rd[14][1].replace('%',''),
            'Yellow cards': rd[15],
            'Red cards': rd[16],
            'Cards per 90': rd[17]
            })

# Convert your list of dictionaries to a DataFrame for processing
df = pd.DataFrame(data)

# Remove % symbol and convert to float
df['Passing accuracy (%)'] = df['Passing accuracy (%)'].str.replace('%', '').astype(float)

# Ensure float type for other columns too
df['Interceptions per 90'] = pd.to_numeric(df['Interceptions per 90'], errors='coerce')
df['Tackles per 90'] = pd.to_numeric(df['Tackles per 90'], errors='coerce')
```

```python
import plotly.express as px
from sklearn.preprocessing import StandardScaler
from sklearn.cluster import KMeans

# Convert percentage strings to float
percent_cols = [
    'Passing accuracy (%)',
    'Long pass accuracy (%)',
    'Successful dribbles per 90 (%)',
    'Aerial duels won per 90 (%)'
]
for col in percent_cols:
    df[col] = df[col].astype(float)

# Select relevant numeric features
numeric_cols = [
    'Key passes per 90',
    'Passing accuracy (%)',
    'Interceptions per 90',
    'Tackles per 90',
    'Balls recovered per game',
    'Total duels won per game',
    'Duels won (%)',
    'Errors leading to goal or shot per 90',
    'Successful dribbles per 90',
    'Successful dribbles per 90 (%)',
    'Aerial duels won per 90 (%)',
    'Cards per 90'
]

numeric_cols_labels = [
    'KPs',
    'Pass(%)',
    'Ints',
    'Tackles',
    'Recs',
    'DWPG',
    'DW(%)',
    'Er90',
    'SucDribs',
    'SucDribs(%)',
    'ADs(%)',
    'Cards'
]

df[numeric_cols] = df[numeric_cols].apply(pd.to_numeric, errors='coerce')

# KMeans clustering
X = df[numeric_cols]
X_scaled = StandardScaler().fit_transform(X)
df['Group'] = KMeans(n_clusters=3, random_state=42).fit_predict(X_scaled)

# Scatter plot with cluster labels

x_variable = 'Balls recovered per game'
y_variable = 'Passing accuracy (%)'

fig_scatter = px.scatter(
    df,
    x=x_variable,
    y=y_variable,
    color='Group',
    #text='Name',
    hover_name='Name',
    title=f'{x_variable} vs {y_variable} Colored by KMeans Cluster',
    color_continuous_scale=px.colors.sequential.Viridis
)

# fig_scatter.update_traces(
#     marker=dict(size=10, opacity=0.8),
#     textposition='top center'  # places names above points
# )

fig_scatter.update_traces(marker=dict(size=10, opacity=0.8))
fig_scatter.update_layout(height=600)
fig_scatter.show()

```

```python
from sklearn.preprocessing import MinMaxScaler

player_names = df['Name']
player_heights = df['Height']
player_cluster = df['Group']

negative_metrics = ['Age','Errors leading to goal or shot','Errors leading to goal or shot per 90','Yellow cards','Red cards','Cards per 90']

metrics_df = df.drop(columns=['Name','Height','Group'])

# Normalize data
scaler = MinMaxScaler()
normalized = scaler.fit_transform(metrics_df)

normalized_df = pd.DataFrame(normalized, columns=metrics_df.columns)

normalized_df[negative_metrics] = -normalized_df[negative_metrics]

# Weights
normalized_df['Age'] *= 5
normalized_df['Minutes played (League)'] *= 5
normalized_df['Key passes per 90'] *= 20
normalized_df['Passing accuracy (%)'] *= 20
normalized_df['Long pass accuracy (%)'] *= 5
normalized_df['Interceptions per 90'] *= 5
normalized_df['Tackles per 90'] *= 5
normalized_df['Balls recovered per game'] *= 20
normalized_df['Total duels won per game'] *= 10
normalized_df['Duels won (%)'] *= 10
normalized_df['Errors leading to goal or shot'] *= 1
normalized_df['Errors leading to goal or shot per 90'] *= 5
normalized_df['Successful dribbles per 90'] *= 1
normalized_df['Successful dribbles per 90 (%)'] *= 5
normalized_df['Aerial duels won per 90'] *= 1
normalized_df['Aerial duels won per 90 (%)'] *= 5
normalized_df['Yellow cards'] *= 1
normalized_df['Red cards'] *= 1
normalized_df['Cards per 90'] *= 1

normalized_df['Name'] = player_names
normalized_df['Height'] = player_heights
normalized_df['Group'] = player_cluster + 1

columns_to_sum = metrics_df.columns
normalized_df['Total score'] = normalized_df[columns_to_sum].sum(axis=1)

# Sort by TotalScore descending
ranked_df = normalized_df.sort_values(by='Total score', ascending=False)

# Reset index and display top players
ranked_df = ranked_df.reset_index(drop=True)

print(ranked_df[['Name', 'Total score'] + list(columns_to_sum)])

fig = px.bar(ranked_df, y='Name', x='Total score', color='Group',title='Player rankings',height=1000)
fig.show()
```
